# Student Spending Analysis

### Python Data Analysis Project

## Project Overview

This project explores monthly spending patterns among six fictional students.

The goal is to analyze expenses, compare individual spending habits, identify major expense categories, and visualize the results using Python.

## Tools and Libraries

- Python
- pandas
- Matplotlib
- Google Colab

## Dataset

The dataset contains six fictional students and five expense categories:

- Rent
- Food
- Transport
- Entertainment
- Shopping

**Note:** All data is synthetic and created for educational purposes.

```python

import pandas as pd

# Monthly expenses of 6 fictional students
data = {
    "Student": ["Alex", "Emma", "John", "Sarah", "David", "Michael"],
    "Rent": [950, 1100, 850, 1000, 900, 1050],
    "Food": [300, 350, 280, 400, 320, 380],
    "Transport": [90, 60, 100, 80, 70, 90],
    "Entertainment": [150, 200, 100, 180, 120, 160],
    "Shopping": [60, 80, 70, 120, 100, 90]
}

df = pd.DataFrame(data)

df
```

```python

# Calculate total expenses for each student

df["Total"] = df[["Rent", "Food", "Transport",
                  "Entertainment", "Shopping"]].sum(axis=1)

df
```

```python
# Basic Statistics

print("Average expenses:", df["Total"].mean())
print("Maximum expenses:", df["Total"].max())
print("Minimum expenses:", df["Total"].min())
```

```python

# Find the student with the highest expenses

highest_index = df["Total"].idxmax()

print("Row index:", highest_index)
print("Student:", df.loc[highest_index, "Student"])
print("Total expenses:", df.loc[highest_index, "Total"])
```

```python
# Find the student with the lowest expenses

lowest_index = df["Total"].idxmin()

print("Row index:", lowest_index)
print("Student:", df.loc[lowest_index, "Student"])
print("Total expenses:", df.loc[lowest_index, "Total"])
```

```python
# Visualize spending by category
import matplotlib.pyplot as plt

# Create a bar chart
plt.bar(df["Student"], df["Total"])

# Add title and labels
plt.title("Monthly Student Expenses")
plt.xlabel("Students")
plt.ylabel("Total Expenses ($)")

# Display the chart
plt.show()
```

```python
# Data Visualization


import matplotlib.pyplot as plt

# Create a bar chart
plt.figure(figsize=(10, 6))

plt.bar(df["Student"], df["Total"], color="green")

# Add title and labels
plt.title("Student Spending Comparison")
plt.xlabel("Students")
plt.ylabel("Total Expenses ($)")

# Add horizontal grid lines
plt.grid(axis="y", linestyle="--", alpha=0.5)

# Display the chart
plt.show()
```

```python

# Spending by category

categories = ["Rent", "Food", "Transport",
              "Entertainment", "Shopping"]

category_totals = df[categories].sum()

print(category_totals)
```

```python
# Category Analysis

categories = ["Rent", "Food", "Transport",
              "Entertainment", "Shopping"]

# Calculate total spending for each category
category_totals = df[categories].sum()

print(category_totals)
```

```python
# Find the highest and lowest spending categories

highest_category = category_totals.idxmax()
lowest_category = category_totals.idxmin()

print("Highest spending category:", highest_category)
print("Lowest spending category:", lowest_category)

print("Highest amount:", category_totals.max())
print("Lowest amount:", category_totals.min())
```

```python

# Visualize spending by category

import matplotlib.pyplot as plt

plt.figure(figsize=(10, 6))

plt.bar(category_totals.index,
        category_totals.values,
        color="skyblue")

plt.title("Total Student Spending by Category")
plt.xlabel("Expense Categories")
plt.ylabel("Total Expenses ($)")

plt.grid(axis="y", linestyle="--", alpha=0.5)

plt.show()
```

```python
import matplotlib.pyplot as plt

# Ensure category_totals is defined in this cell
categories = ["Rent", "Food", "Transport", "Entertainment", "Shopping"]
category_totals = df[categories].sum()

# Create a bar chart
plt.figure(figsize=(10, 6))

plt.bar(category_totals.index,
        category_totals.values,
        color="orange")

# Add values above each bar
for i, value in enumerate(category_totals.values):
    plt.text(i, value + 50, f"${value:,}",
             ha="center")

# Add title and labels
plt.title("Total Student Spending by Category")
plt.xlabel("Expense Categories")
plt.ylabel("Total Expenses ($)")

# Add grid
plt.grid(axis="y", linestyle="--", alpha=0.5)

# Display chart
plt.show()
```

```python
# Percentage of Total Spending

# Calculate total spending
total_spending = category_totals.sum()

# Calculate percentage for each category
category_percentages = (category_totals / total_spending) * 100

print("Total spending:", total_spending)
print("\nSpending percentages:")
print(category_percentages.round(2))
```

```python
# Create a pie chart

import matplotlib.pyplot as plt

plt.figure(figsize=(8, 8))

plt.pie(
    category_totals.values,
    labels=category_totals.index,
    autopct="%1.1f%%",
    startangle=90
)

plt.title("Student Spending Distribution")

plt.show()
```
